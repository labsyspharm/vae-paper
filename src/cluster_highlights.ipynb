```python
import os

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# I/O
main = pd.read_csv(os.path.join(os.getcwd(), 'input/main.csv'))

out = os.path.join(os.getcwd(), 'output/cluster_highlights')
if not os.path.exists(out):
    os.makedirs(out)
```

```python
# plot
for clustering, emb in zip(['Seg', 'VAE9_VIG7', 'VAE30'], ['Seg_emb', 'VAE9_VIG7_emb', 'VAE30_emb']): 

    data = main.sample(frac=0.05)
    
    for i in sorted(data[clustering].unique()):
        if i != -1:
            
            print(f'{clustering} cluster {i}')
            
            fig = plt.figure()
            ax = plt.axes(projection='3d', computed_zorder=False)
            
            if clustering == 'Seg':
                ax.view_init(elev=10, azim=230)
            elif clustering == 'VAE9_VIG7':
                ax.view_init(elev=10, azim=236)
            elif clustering == 'VAE30':
                ax.view_init(elev=10, azim=240)

            ax.scatter(
                data[f'{emb}1'],
                data[f'{emb}2'],
                data[f'{emb}3'],
                c='gainsboro', linewidth=0.0, s=2, alpha=0.5
            )
            ax.scatter(
                data[f'{emb}1'][data[clustering] == i],
                data[f'{emb}2'][data[clustering] == i],
                data[f'{emb}3'][data[clustering] == i],
                color=(0, 0.33, 1), linewidth=0.0, s=3, alpha=1,
            )
            ax.set_xlabel('UMAP1')
            ax.set_ylabel('UMAP2')
            ax.set_zlabel('UMAP3')

            ax.axis('auto')
            ax.tick_params(labelsize=10)
            ax.grid(True)
            plt.tight_layout()
            
            plt.savefig(os.path.join(out, f'{clustering}_cluster_{i}.png'), dpi=800)
            
            plt.show()
```
